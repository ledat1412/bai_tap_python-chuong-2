```python
##Bài 2.3 TH 2.3.

Viết chương trình tính lương tuần như Ví dụ 2.1 (lương 100000 đồng/giờ, giờ vượt 40 tính gấp 1,5) nhưng nhập lại cho đến khi người dùng gõ một số không âm: chữ hoặc số âm đều phải báo lỗi và hỏi lại. Kiểm tra với dãy gõ abc, -2, 45
```

```python
while True:
    try:
        gio = float(input("Nhập số giờ làm: "))
        if gio < 0:
            print("Lỗi: Không được nhập số âm. Hãy nhập lại!")
        else:
            break
    except:
        print("Lỗi: Phải nhập số, không được nhập chữ. Hãy nhập lại!")

if gio <= 40:
    luong = gio * 100000
else:
    luong = 40 * 100000 + (gio - 40) * 150000

print("Tổng lương tuần là:", luong)
```

```python
#bài2.4 Nhập liên tục các số thực (bỏ qua và báo giá trị không phải số) cho đến khi gõ xong. In số giá trị hợp lệ, tổng, trung bình, giá trị lớn nhất và nhỏ nhất; nếu chưa nhập số nào thì báo “chưa có dữ liệu”. Kiểm tra với dãy gõ: 7.5, 8, abc, 6.5, 9
```

```python
ds = []

while True:
    n = input("Nhập số (hoặc gõ 'xong'): ")
    if n == 'xong':
        break
    try:
        ds.append(float(n))
    except:
        print("Không phải số, bỏ qua!")

if len(ds) == 0:
    print("chưa có dữ liệu")
else:
    tong = sum(ds)
    sl = len(ds)
    
    print("Số giá trị hợp lệ:", sl)
    print("Tổng:", tong)
    print("Trung bình:", tong / sl)
    print("Lớn nhất:", max(ds))
    print("Nhỏ nhất:", min(ds))
```

```python
n = 27
buoc = 0
lon_nhat = n

while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1
        
    buoc += 1
    if n > lon_nhat:
        lon_nhat = n

print("Số bước:", buoc)
print("Giá trị lớn nhất:", lon_nhat)
```

```python
import random
import math

def uoc_luong(N: int) -> float:
    M = 0

    for i in range(N):
        x = random.random()
        y = random.random()

        if x**2 + y**2 <= 1:
            M += 1

    return 4 * M / N


random.seed(2026)

for N in [10**2, 10**4, 10**6]:
    pi_uoc_luong = uoc_luong(N)
    sai_so = abs(pi_uoc_luong - math.pi)

    print(f"N = {N}")
    print(f"Ước lượng π = {pi_uoc_luong}")
    print(f"Sai số tuyệt đối = {sai_so}")
    print(f"Ngưỡng 6/N = {6/N}")

    assert sai_so < 6 / math.sqrt(N)
```

```python
def f(x: float) -> float:
    return x**3 - x - 2

def chia_doi(a: float, b: float, sai_so: float) -> None:
    buoc = 0
    while b - a >= sai_so:
        c = (a + b) / 2

        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

        buoc += 1

    nghiem = (a + b) / 2
    ly_thuyet = math.log2(1 / sai_so)

    print("Nghiệm gần đúng:", nghiem)
    print("Số bước thực tế:", buoc)
    print("Số bước lý thuyết:", ly_thuyet)


chia_doi(1, 2, 10**-10)
```
